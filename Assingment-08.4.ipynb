```python
X = df[["Income", "Credit_Score", "Loan_Amount"]]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
# task-04
```

```python
hc = AgglomerativeClustering(
    n_clusters=5
)
clusters = hc.fit_predict(X_scaled)
df['Cluster']=clusters
df.head()
```

```python
# Task-05
```

```python
df["n_clusters"] = hc.fit_predict(X_scaled)
print(df.head(10))
print(df["n_clusters"].value_counts().sort_index())
```
